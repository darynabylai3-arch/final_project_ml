# Random Forest

Ensemble of decision trees trained on bootstrap samples. Captures non-linear effects and interactions without scaling.

The pipeline performs median imputation, binary imputation (most frequent value), one-hot encoding of `cp`, `restecg` and `slope`, and no scaling (not needed for tree models). All fitting happens inside cross-validation folds and on the training set only.
