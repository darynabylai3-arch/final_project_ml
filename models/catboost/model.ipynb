# CatBoost

Gradient boosting over decision trees. Used as the model behind the web service in the earlier project.

The pipeline performs median imputation, binary imputation (most frequent value), one-hot encoding of `cp`, `restecg` and `slope`, and no scaling (not needed for tree models). All fitting happens inside cross-validation folds and on the training set only.

```python
import sys
from pathlib import Path
sys.path.append(str(Path('../../src').resolve()))
from evaluation import evaluate
```

```python
from catboost import CatBoostClassifier
```

```python
metrics = evaluate(
    name='CatBoost',
    estimator=CatBoostClassifier(iterations=500, learning_rate=0.05, depth=6, random_seed=42, verbose=0, allow_writing_files=False),
    scale=False,
    out_dir=Path('.'),
)
metrics
```
